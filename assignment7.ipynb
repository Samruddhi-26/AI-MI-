```python
# 8.Calculate the average marks in each subject (column-wise mean).
import numpy as np

arr = np.array([
    [101, 85, 90, 78],
    [102, 75, 88, 92],
    [103, 95, 76, 85],
    [104, 80, 82, 89]
])

avg = np.mean(arr[:, 1:], axis=0)

print("Average marks in each subject:", avg)
```

```python
# 9.Get the data of students who scored at least 80 in both subjects.
import numpy as np

arr = np.array([
    [101, 85, 90, 78],
    [102, 75, 88, 92],
    [103, 95, 76, 85],
    [104, 80, 82, 89]
])

result = arr[np.all(arr[:, 2:] >= 80, axis=1)]

print(result)
```

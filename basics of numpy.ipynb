```python
# NumPy Searching and Filtering Methods
# 1.where
a = np.array([10, 20, 30, 40, 50])

b = np.where(a > 25)

print(b)
```

```python
# 2.argmax()
a = np.array([10, 50, 30, 80, 20])

print(np.argmax(a))
```

```python
# 3.argmin()
a = np.array([10, 50, 30, 80, 20])

print(np.argmin(a))
```
